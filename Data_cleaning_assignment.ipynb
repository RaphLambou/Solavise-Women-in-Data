### Filling missing data in the gender and age range columns

```python
age_missing = df[df['Age Range'].isnull()]
age_missing
```

```python
# extracting the age and sex from the findings column
import re

def age_range(age):
    if age < 11:
        return '0-10'
    elif age < 21:
        return '11-20'
    elif age < 31:
        return '21-30'
    elif age < 41:
        return '31-40'
    elif age < 51:
        return '41-50'
    elif age < 65:
        return '51-64'
    else:
        return '65+'

def pat_gender(text):
    text = text.lower()
    if text == 'male' or text == 'man':
        return 'Male'
    elif text =='female' or text == 'woman':
        return 'Female'
    else:
        return None

def fill_missing_age_sex(text):
    match = re.search(r'(\d+)[-\s]*year[-\s]*old\s+(male|female|man|woman)', text, re.IGNORECASE)
    if match:
        age = age_range(int(match.group(1))) if match.group(1) else None
        gender = pat_gender(match.group(2)) if match.group(2) else None
        return age, gender


df[['Age Range', 'Patient Gender']] = df.apply(lambda row: fill_missing_age_sex(row['Findings']) if pd.isnull(row['Age Range']) or pd.isnull(row['Patient Gender']) else (row['Age Range'], row['Patient Gender']), axis=1, result_type='expand')
        
        
```

```python
def fill_missing_gender(text):
    match = re.search(r'(female|male)', text, re.IGNORECASE)
    if match:
        gender = pat_gender(match.group(1)) if match.group(1) else None
        return gender
    
df['Patient Gender'] = df.apply(lambda row: fill_missing_gender(row['Findings']) if pd.isnull(row['Patient Gender']) else row['Patient Gender'], axis=1)
```

```python
def fill_missing_age(text):
    match = re.search(r'(\d+)[-\s]*year[-\s]*old', text, re.IGNORECASE)
    if match:
        age = age_range(int(match.group(1))) if match.group(1) else None
        return age  

df['Age Range'] = df.apply(lambda row: fill_missing_age(row['Findings']) if pd.isnull(row['Age Range']) else row['Age Range'], axis=1)
```
